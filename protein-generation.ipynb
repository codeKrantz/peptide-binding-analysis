```python
import csv
import itertools
```

```python
AMINO_ACIDS = list("ACDEFGHIKLMNPQRSTVWY")
_VALID_AA = set(AMINO_ACIDS)

def is_valid_sequence(seq: str) -> bool:
    """Return True only if seq is non-empty and every character is a standard amino acid."""
    return bool(seq) and all(c in _VALID_AA for c in seq)
```

# 1 aa (20 cases)

```python
valid_sequences = []

for aa in AMINO_ACIDS:
    valid = is_valid_sequence(aa)
    print(f"{aa}: {'valid' if valid else 'INVALID'}")
    if valid:
        valid_sequences.append(aa)

output_path = "sequences_1aa.csv"
with open(output_path, "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["sequence"])
    for seq in valid_sequences:
        writer.writerow([seq])

print(f"\nWrote {len(valid_sequences)} valid sequences to {output_path}")
```

# 2 aa (400 cases)

```python
valid_sequences_2aa = []

for aa1, aa2 in itertools.product(AMINO_ACIDS, repeat=2):
    seq = aa1 + aa2
    valid = is_valid_sequence(seq)
    print(f"{seq}: {'valid' if valid else 'INVALID'}")
    if valid:
        valid_sequences_2aa.append(seq)

output_path = "sequences_2aa.csv"
with open(output_path, "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["sequence"])
    for seq in valid_sequences_2aa:
        writer.writerow([seq])

print(f"\nWrote {len(valid_sequences_2aa)} valid sequences to {output_path}")
```

# Combining Sequences

```python
import time

input_files = [
    "sequences_1aa.csv",
    "sequences_2aa.csv",
    "sequences_3aa.csv",
    "sequences_4aa.csv",
    "sequences_5aa.csv",
    "sequences_6aa.csv",
]
output_path = "sequences_all.csv"

start = time.time()
total = 0

with open(output_path, "w", newline="") as out:
    for i, path in enumerate(input_files):
        with open(path, "r", newline="") as f:
            header = f.readline()      # consume header
            if i == 0:
                out.write(header)      # write header once
            for line in f:
                out.write(line)
                total += 1

elapsed = time.time() - start
print(f"Wrote {total:,} sequences to {output_path} in {elapsed:.1f}s")
```

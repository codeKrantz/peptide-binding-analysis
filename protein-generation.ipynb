```python
import csv
import itertools
```

# 1 aa (20 cases)

```python
AMINO_ACIDS = list("ACDEFGHIKLMNPQRSTVWY")

sequences_1aa = [aa for aa in AMINO_ACIDS]

for seq in sequences_1aa:
    print(seq)

output_path = "sequences_1aa.csv"
with open(output_path, "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["sequence"])
    for seq in sequences_1aa:
        writer.writerow([seq])

print(f"\nWrote {len(sequences_1aa)} sequences to {output_path}")
```

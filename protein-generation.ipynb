```python
import csv
import itertools
```

```python
AMINO_ACIDS = list("ACDEFGHIKLMNPQRSTVWY")
_VALID_AA = set(AMINO_ACIDS)

def is_valid_sequence(seq: str) -> bool:
    """Return True only if seq is non-empty and every character is a standard amino acid."""
    return bool(seq) and all(c in _VALID_AA for c in seq)
```

# 1 aa (20 cases)

```python
valid_sequences = []

for aa in AMINO_ACIDS:
    valid = is_valid_sequence(aa)
    print(f"{aa}: {'valid' if valid else 'INVALID'}")
    if valid:
        valid_sequences.append(aa)

output_path = "sequences_1aa.csv"
with open(output_path, "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["sequence"])
    for seq in valid_sequences:
        writer.writerow([seq])

print(f"\nWrote {len(valid_sequences)} valid sequences to {output_path}")
```
